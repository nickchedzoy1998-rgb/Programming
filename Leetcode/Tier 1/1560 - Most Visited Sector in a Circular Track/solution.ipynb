```python
class Solution(object):

    def mostVisited(self, n, rounds):
        start = rounds[0]
        end = rounds[-1]

        if start == end:
            return [start]

        elif start < end:
            return [x for x in range(start, n+1) if x <= end]

        else:
            return [x for x in range(1, n+1) if x>= start or x<=end]
```

```python
example_1 = Solution()

example_1.mostVisited(n=4, rounds=[1,3,1,2])
```
