```python
class Solution(object):
    def findNumbers(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        count = 0

        for num in nums:
            if len(str(num)) % 2 == 0:
                count += 1

        return count
```

```python
example_1 = Solution()
result_1 = example_1.findNumbers(nums = [12,345,2,6,7896])

expected_answer_1 = 2


if result_1 == expected_answer_1:
    print('Correct!')
else:
    print('Incorrect :(')
```
