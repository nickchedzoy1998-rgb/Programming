```python
# Example
# Input: nums = [1,1,2]
# Output: 2, nums = [1,2,_]
```

```python
def removeDuplicates(nums):
    nums = sorted(nums)
    final_list = []

    for num in nums:
        if len(final_list) == 0:
            final_list.append(num)
        else:
            if num == final_list[-1]:
                continue
            else:
                final_list.append(num)

    final_list_len = len(final_list)

    print(f'{final_list_len}, nums = {final_list}')
```

```python
nums = [1,1,2]
removeDuplicates(nums)
```

```python
# Input: nums = [0,0,1,1,1,2,2,3,3,4]
# Output: 5, nums = [0,1,2,3,4,_,_,_,_,_]

nums = [0,0,1,1,1,2,2,3,3,4]

removeDuplicates(nums)
```

```python
# Must Return in Place and Return Integer

def removeDuplicates(nums):
    
    """
    :type nums: List[int]
    :rtype: int
    """
    write = 1

    for read in range(1, len(nums)):
        if nums[read] != nums[read - 1]:
            nums[write] = nums[read]
            write += 1

    return write
```

```python
nums = [1,1,2]
removeDuplicates(nums)
```

```python
# Input: nums = [0,0,1,1,1,2,2,3,3,4]
# Output: 5, nums = [0,1,2,3,4,_,_,_,_,_]

nums = [0,0,1,1,1,2,2,3,3,4]

removeDuplicates(nums)
```
