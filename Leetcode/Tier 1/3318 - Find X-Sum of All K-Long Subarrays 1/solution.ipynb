```python
from collections import Counter
```

```python
class Solution(object):
    def findXSum(self, nums, k, x):
        """
        :type nums: List[int]
        :type k: int
        :type x: int
        :rtype: List[int]
        """

        start = 0
        end = start + k

        answer = []

        while end <= len(nums):
            current_array = [num for num in nums[start: end]]

            frequencies = Counter(current_array)
            sorted_frequencies = sorted(frequencies.items(), key=lambda item: (item[1], item[0]), reverse=True)

            keep = sorted_frequencies[:x]

            array_sum = sum(num*freq for num, freq in keep)

            answer.append(array_sum)

            start += 1
            end += 1

        return answer
```

```python
example_1 = Solution()
example_1. findXSum(nums = [1,1,2,2,3,4,2,3], k = 6, x = 2)
```
