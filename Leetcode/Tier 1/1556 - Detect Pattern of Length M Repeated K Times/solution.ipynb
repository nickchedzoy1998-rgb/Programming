```python
"""
Leetcode.Easy.1556 - Detect Pattern of Length M Repeated K Times.solution.ipynb
Given an array of positive integers arr, find a pattern of length m that is repeated k or more times.

A pattern is a subarray (consecutive sub-sequence) that consists of one or more values, 
repeated multiple times consecutively without overlapping. A pattern is defined by its length and the number of repetitions.

Return true if there exists a pattern of length m that is repeated k or more times, otherwise return false.
"""
```

```python
class Solution:
    def containsPattern(self, arr: list[int], m: int, k: int) -> bool:

        for i in range(0, m):
            inner_list = []
            start = i
            end = i + m
            prev_list = []
            current_match_count = 1
            while end <= len(arr):

                for x in range(0, (int(len(arr)/m))):    
                    for num in arr[start:end]:
                        
                        inner_list.append(num)
                        if len(inner_list) == m:
                            start += m
                            end += m
                            
                            if prev_list == inner_list:
                                current_match_count +=1
                            else:
                                current_match_count = 1

                            if current_match_count == k:
                                return True
                            prev_list = inner_list.copy()
                            inner_list = []
                        
        return False
```

```python
tick = '\U00002705'
cross = '\U0000274C'

#Example 1

arr = [1,2,4,4,4,4] 
m = 1 
k = 3
answer = True

q1 = Solution()
if q1.containsPattern(arr, m, k) == answer:
    print(f'Correct! {tick}')
else:
    print(f'Incorrect :( {cross}')
```
