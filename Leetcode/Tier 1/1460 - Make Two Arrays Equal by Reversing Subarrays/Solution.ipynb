```python
class Solution(object):
    def canBeEqual(self, target, arr):
        """
        :type target: List[int]
        :type arr: List[int]
        :rtype: bool
        """
        if sorted(target) == sorted(arr):
            return True

        return False
```

```python
example_1 = Solution()
answer_1 = True

if (example_1.canBeEqual([1,2,3,4], [2,4,1,3])) == answer_1:
    print('Correct!')
else:
    print('Incorrect :(')
```
