```python
class Solution(object):
    def minOperations(self, logs):
        """
        :type logs: List[str]
        :rtype: int
        """
        required_ops = 0

        for log in logs:
            if log == "./" or (log == "../" and required_ops < 1):
                continue

            elif log == "../":
                required_ops -= 1              
            
            else:
                required_ops += 1

        return required_ops
```

```python
example1 = Solution()
example1.minOperations(logs = ["d1/","d2/","../","d21/","./"])
```

```python
example2 = Solution()
example2.minOperations(logs = ["d1/","d2/","./","d3/","../","d31/"])
```
