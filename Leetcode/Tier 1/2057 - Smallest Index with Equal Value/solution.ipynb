```python
class Solution(object):
    def smallestEqual(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """

        for i, num in enumerate(nums):
            if (i%10) == num:
                return i
        
        return -1
```

```python
example_1 = Solution()
example_1.smallestEqual([0,1,2])
```

```python
example_2 = Solution()
example_2.smallestEqual([4,3,2,1])
```

```python
example_3 = Solution()
example_3.smallestEqual([1,2,3,4,5,6,7,8,9,0])
```
