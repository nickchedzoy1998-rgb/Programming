```python
class Solution(object):
    def countSubarrays(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        count = 0
        
        for i in range(0, len(nums)-2):
            sub_array = nums[i:i+3]

            if (sub_array[0] + sub_array[2])*2 == sub_array[1]:
                count += 1

        return count
```

```python
example_1 = Solution()
answer_1 = 1

if example_1.countSubarrays([1,2,1,4,1]) == answer_1:
    print('Correct!')
else:
    print('Incorrect :(')
```
