```python
def findTheDistanceValue(arr1: list[int], arr2: list[int], d: int) -> int:

    low_dif_count = 0
    nums_to_check = len(arr1)

    for num_a1 in arr1:
        for num_a2 in arr2:
            if num_a1 == num_a2:
                low_dif_count +=1
                break
            if num_a1 > num_a2:
                dif = num_a1 - num_a2
            else:
                dif = num_a2 - num_a1
            if dif > d:
                continue
            else:
                low_dif_count +=1
                break
    
    distance = nums_to_check - low_dif_count
    return distance
```

```python
arr1 = [4,5,8]
arr2 = [10,9,1,8]
d = 2
answer = 2
result = findTheDistanceValue(arr1, arr2, d)

if answer == result:
    print (f'Answer: {answer}\nYour Answer: {result}\nCorrect!')
else:
    print('X INCORRECT')
```
