```python
def halvesAreAlike(s: str) -> bool:
    string_length = len(s)

    a_index = string_length/2
    a = []
    b = []

    vowels = ['a', 'e', 'i', 'o', 'u']

    for i, char in enumerate(s):
        if i < a_index:
            a.append(char)
        else:
            b.append(char)

    a_vowel_count = 0
    b_vowel_count = 0

    for letter in a:
        if str.lower(letter) in vowels:
            a_vowel_count +=1
    for letter in b:
        if str.lower(letter) in vowels:
            b_vowel_count +=1

    return a_vowel_count == b_vowel_count
```

```python
halvesAreAlike('book')
```

```python
halvesAreAlike('textbook')
```

```python
halvesAreAlike('Uo')
```
