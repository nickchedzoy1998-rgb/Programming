```python
class Solution:
    def separateDigits(self, nums: list[int]) -> list[int]:
        
        output_list = []

        for num in nums:
            if num > 9:
                for digit in str(num):
                    convert_back = output_list.append(int(digit))
            else:
                output_list.append(num)
            
        return output_list
```

```python
input_1 = Solution()
answer_1 = [1,3,2,5,8,3,7,7]
output_1 = input_1.separateDigits(nums=[13,25,83,77])

if output_1 == answer_1:
    print('Correct! :)')
else:
    print('Incorrect :(')
```
