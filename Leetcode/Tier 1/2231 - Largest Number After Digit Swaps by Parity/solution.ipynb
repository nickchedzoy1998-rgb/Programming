```python
class Solution(object):
    def largestInteger(self, num):
        """
        :type num: int
        :rtype: int
        """
        num_string = str(num)

        evens = sorted([int(digit) for digit in num_string if int(digit) % 2 == 0], reverse=True)
        evens_index = sorted([i for i, digit in enumerate(num_string) if int(digit) % 2 == 0])
        evens_new_idx = list(zip(evens, evens_index))

        odds = sorted([int(digit) for digit in num_string if int(digit) % 2 == 1], reverse=True)
        odds_index = sorted([i for i, digit in enumerate(num_string) if int(digit) % 2 == 1])
        odds_new_idx = list(zip(odds, odds_index))

        combined = odds_new_idx + evens_new_idx

        sorted_combined = sorted(combined, key=lambda x: x[1])

        combined_sorted_list = [tpl[0] for tpl in sorted_combined]

        return int("".join(str(d) for d in combined_sorted_list))
```

```python
example_1 = Solution()

example_1.largestInteger(1234)
```
