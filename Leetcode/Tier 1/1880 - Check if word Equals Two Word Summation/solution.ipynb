```python
import string

class Solution(object):
    def isSumEqual(self, firstWord, secondWord, targetWord):
        """
        :type firstWord: str
        :type secondWord: str
        :type targetWord: str
        :rtype: bool
        """
        alpha_idx = {letter: idx for idx, letter in enumerate(string.ascii_lowercase)}

        def get_value(word):
            value_str = ''
            idx_list = [alpha_idx.get(char) for char in word]
            value_str += "".join(str(idx) for idx in idx_list)

            return int(value_str)
        
        return (get_value(firstWord) + get_value(secondWord)) == get_value(targetWord)
```

```python
example_1 = Solution()

example_1.isSumEqual('acb', 'cba', 'cdb')
```

```python
example_2 = Solution()

example_2.isSumEqual('aaa', 'a', 'aab')
```
