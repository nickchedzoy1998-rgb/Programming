```python
from itertools import combinations

class Solution(object):
    def countPairs(self, nums, k):
        """
        :type nums: List[int]
        :type k: int
        :rtype: int
        """
        pairs = 0

        for (idx1, val1), (idx2, val2) in combinations(enumerate(nums), 2):
            if val1 != val2:
                continue

            if (idx1 * idx2) % k != 0:
                continue

            else:
                pairs += 1

        return pairs
```

```python
example_1 = Solution()
example_1.countPairs(nums = [3,1,2,2,2,1,3], k = 2)
```

```python
example_2 = Solution()
example_2.countPairs(nums = [1,2,3,4], k = 1)
```
