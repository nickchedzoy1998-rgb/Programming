```python
from collections import Counter
```

```python
class Solution(object):
    def uncommonFromSentences(self, s1, s2):
        """
        :type s1: str
        :type s2: str
        :rtype: List[str]
        """
        combined_list = (s1.split(" ")) + (s2.split(" "))
        freq = dict(Counter(combined_list))

        uncommon = []

        for word in combined_list:
            if freq.get(word) == 1:
                uncommon.append(word)

        return uncommon
```

```python
example_1 = Solution()

example_1.uncommonFromSentences(s1 = "this apple is sweet", s2 = "this apple is sour")
```

```python
example_2 = Solution()

example_2.uncommonFromSentences(s1 = "apple apple", s2 = "banana")
```
