```python
# class Solution:

def isValid(s:str) -> bool:

    openers = ['(', '[', '{']
    closers = [')', ']', '}']

    open_to_close = dict(zip(openers, closers))

    open_brackets = []

    for bracket in s:
        if bracket in openers:
            open_brackets.append(bracket)

        else:
            if len(open_brackets) == 0:
                return False
            
            last_opened_bracket = open_brackets.pop()
            closing_bracket_needed = open_to_close[last_opened_bracket]
            if bracket != closing_bracket_needed:
                return False
            
    if len(open_brackets) == 0:
        return True
```

```python
# Example 1:
# Output: true

isValid("()")
```

```python
# # Example 2:
# Output: true
isValid("()[]{}")
```

```python
# # Example 3:
# Output: false
isValid("(]")
```

```python
# # Example 4:
# Output: true
isValid("([])")
```

```python
# # Example 5:
# Output: false
isValid("([)]")
```
