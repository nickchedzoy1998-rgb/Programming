```python
import itertools
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def hasPathSum(self, root: TreeNode, targetSum: int) -> bool:
        if not root:
            return False
        
        def count_layers(root):
            if not root:
                return 0
            
            queue = [root]
            layers = 0
            
            while queue:
                layers += 1
                next_layer = []
                for node in queue:
                    if node.left: 
                        next_layer.append(node.left)
                        # print(f'LEFT: {node.left.val} added to next layer')
                    if node.right: 
                        next_layer.append(node.right)
                        # print(f'RIGHT: {node.right.val} added to next layer')
                queue = next_layer
                
            return layers - 1
        
        directions = ['left', 'right']
        search_combinations = list(itertools.product(directions, repeat=count_layers(root)))

        for search in search_combinations:
            values = [root.val]
            current_position = root
            for dir in search:
                if dir == 'left':
                    if current_position.left:
                        current_position = current_position.left
                        values.append(current_position.val)
                    else: break 
                if dir == 'right':
                    if current_position.right:
                        current_position = current_position.right
                        values.append(current_position.val)
                    else: break

            if current_position.left is None and current_position.right is None:
                if sum(values) == targetSum:
                    return True

        return False 
```

```python
### --- Example 1 --- ### 

# Leaves
leaf_7 = TreeNode(7)
leaf_2 = TreeNode(2)
leaf_13 = TreeNode(13)
leaf_1 = TreeNode(1)

# Middle level
node_11 = TreeNode(11, left=leaf_7, right=leaf_2)
node_4_right = TreeNode(4, right=leaf_1)
node_8 = TreeNode(8, left=leaf_13, right=node_4_right)

# Upper level
node_4_left = TreeNode(4, left=node_11)

# Root
root = TreeNode(5, left=node_4_left, right=node_8)

solution_1 = Solution()
correct_answer = True

if solution_1.hasPathSum(root, 22) == correct_answer:
    print('Correct!')
else:
    print('Incorrect :(')
```

!* Above Class Fails on Memory Limit *!

```python
class Solution:
    def hasPathSum(self, root: TreeNode, targetSum: int) -> bool:
        if not root:
            return False
        
        bookmarks = [(root, root.val)]

        while bookmarks:
            current_position, current_tot = bookmarks.pop()
            
            if not current_position.left and not current_position.right:
                if current_tot == targetSum:
                    return True
            
            if current_position.right:
                bookmarks.append((current_position.right, current_tot + current_position.right.val))
            
            if current_position.left:
                bookmarks.append((current_position.left, current_tot + current_position.left.val))
        
        return False
```
