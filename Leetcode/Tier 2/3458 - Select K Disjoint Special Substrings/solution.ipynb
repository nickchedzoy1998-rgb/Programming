```python
from collections import Counter
```

```python
class Solution(object):
    def maxSubstringLength(self, s, k):
        """
        :type s: str
        :type k: int
        :rtype: bool
        """
        if k==0:
            return True
        
        disjoints = 0
        letter_counts = Counter(s)
        unique_letters = [l for l, count in letter_counts.items() if count == 1]

        for char in s:
            i = s.index(char)
            next_index = i+1
            while next_index < len(s):
                if s[next_index] in unique_letters:
                    disjoints += 1
                    next_index +=1
                else:
                    break
                if disjoints >= k:
                    return True

        return False
```

```python
s_1 = "abcdbaefab"
example_1 = Solution()

answer_1 = True

if example_1.maxSubstringLength(s=s_1, k=2) == answer_1:
    print('CORRECT!')
else:
    print('Incorrect :(')
```

```python
class Solution(object):
    def maxSubstringLength(self, s, k):
        """
        :type s: str
        :type k: int
        :rtype: bool
        """
        if k==0:
            return True
        
        disjoints = 0

        # Identify first & last index of each letter
        letter_indexes = {}

        for i, char in enumerate(s):
            if char not in letter_indexes:
                letter_indexes[char] = {'first': i, 'last': i}
            else:
                letter_indexes[char]['last'] = i

        intervals = []

        for i, char in enumerate(s):
            if letter_indexes.get(char).get('first') != i:
                continue

            start = i
            end = letter_indexes.get(char).get('last')
            j=start
            valid = True

            while j <= end:
                c = s[j]

                if letter_indexes.get(c).get('first') < start:
                    valid = False
                    break

                if letter_indexes.get(c).get('last') > end:
                    end = letter_indexes.get(c).get('last')
                
                j+=1

            if end - start == len(s) - 1:
                valid = False

            if valid == True:
                intervals.append((start, end))

        intervals_sorted = sorted(intervals, key=lambda x: x[1])

        last_end = -1

        for start, end in intervals_sorted:
            if start > last_end:
                disjoints += 1
                last_end = end
                if disjoints >= k:
                    return True

        return False
```

```python
s_4 = "wwo"
example_4 = Solution()

answer_4 = True

if example_4.maxSubstringLength(s=s_4, k=1) == answer_4:
    print('CORRECT!')
else:
    print('Incorrect :(')
```
