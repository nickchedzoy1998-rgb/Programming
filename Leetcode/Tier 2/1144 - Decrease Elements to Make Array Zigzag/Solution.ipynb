```python
class Solution(object):
    def movesToMakeZigzag(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        even_index_list = [i for i in range(0, len(nums)) if i % 2 == 0]
        odd_index_list = [i for i in range(0, len(nums)) if i not in even_index_list]

        even_moves = 0
        odd_moves = 0

        original_list = nums.copy()

        # EVEN
        for idx in even_index_list:
            if idx == 0:
                adj_left = 1
            else:
                adj_left = nums[idx] - nums[idx-1]

            if adj_left < 1:
                moves = (abs(nums[idx] - nums[idx-1])) + 1
                even_moves += moves
                nums[idx-1] = nums[idx-1] - moves

            if idx == (len(nums)-1):
                adj_right = 1
            else:
                adj_right = nums[idx] - nums[idx+1]

            if adj_right < 1:
                moves = (abs(nums[idx] - nums[idx+1])) + 1
                even_moves += moves
                nums[idx+1] = nums[idx+1] - moves

        if even_moves == 0:
            return even_moves

        nums = original_list

        # ODD
        for idx in odd_index_list:
            adj_left = nums[idx] - nums[idx-1]

            if adj_left < 1:
                moves = (abs(nums[idx] - nums[idx-1])) + 1
                odd_moves += moves
                nums[idx-1] = nums[idx-1] - moves

            if idx == (len(nums)-1):
                adj_right = 1
            else:
                adj_right = nums[idx] - nums[idx+1]

            if adj_right < 1:
                moves = (abs(nums[idx] - nums[idx+1])) + 1
                odd_moves += moves
                nums[idx+1] = nums[idx+1] - moves

            if odd_moves == even_moves:
                return even_moves

        return min(even_moves, odd_moves)
```

```python
example_1 = Solution()
result_1 = example_1.movesToMakeZigzag(nums=[1,2,3])

expected_result_1 = 2

if result_1 == expected_result_1:
    print('Correct!')
else:
    print('False :(')
```
